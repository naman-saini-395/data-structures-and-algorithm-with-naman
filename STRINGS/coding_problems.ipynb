1 . count vowels and consonents

assume all characters are lowercase

string may include speical characters

```python
s = '_hel lo 123$world'

vowels = 0
consonents = 0
for ch in s:
    if ch.isalpha():
        if ch in 'aeiou':
            vowels += 1
        else:
            consonents +=1
            
print(f'There are {vowels} vowels and {consonents} consonents')
```

```python
# 3. palindrome string 

s = 'naman'

s_rev = s[::-1]
if s == s_rev:
    print('Pallindrome')
else:
    print('Not Palindrome')
```

```python
# # 4. Check for Anagrams
# return ture if two string contain the same characters in any any order
# string will be of same length and lowercase

s1 = 'listen'
s2 = 'silent'
```

```python
# Time nlog(n)
s1_sorted = sorted(s1)
s2_sorted = sorted(s2)

if s1_sorted == s2_sorted:
    print('Anagram')
else:
    print('not Anagram')
```

```python
mapping = [0] * 26

for ch in s1:
    index = ord(ch) - 97
    # ptrint(ch, ord(ch), index)
    mapping[index] +=1
    
mapping
for ch in s2:
    index = ord(ch) - 97
    mapping[index] -= 1
if all(m == 0 for m in mapping):
    print("Anagram")
else:
    print('not Anagram')
```
