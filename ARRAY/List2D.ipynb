```python
# ceration
matrix =[
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print(type(matrix))
```

```python
#display
for row in matrix:
    print(row)
```

```python
def display_matrix(mat):
    for row in mat:
        print(row)
```

```python
display_matrix(matrix)
```

```python
#traversal

for row in matrix:
    for ele in row:
        print(ele , end=" ")
    print()
```

```python
m = len(matrix)
n = len(matrix[0])


for i in range(n):
    for j in range(n):
        print(matrix[i][j], end=" ")
    print()
```
