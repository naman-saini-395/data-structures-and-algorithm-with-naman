```python
# 1 find the lagerst no in a ;list


numbers = [12, 45, 7, 89, 34]

largest = numbers[0]
for element in numbers[1:]:
    if element > largest:
        largest = element
        
        
print(f'largest element: {largest}')
    
```

```python
# 2 count occurrences of an Element 

items = ["apple", "banana", "apple", "orange", "banana", "apple"]
target = "apple" 

# items.count(target)
# o(n)

count = 0
for item in items:
    if item == target:
        count += 1

print(f'{target} appears {count} time ')
```

```python
# Remove Duplicates form a list 

numbers = [1, 2, 2, 3, 4, 4, 5]

# list(set(numbers))

#o(n)
 
unique_element = []
for number in numbers:
    if number not in unique_element:
        unique_element.append(number)
print(f'unique_element:',unique_element)
        



```

```python
# reverse a list ( without the reverse mathod)

data = [4, 6, 2, 3, 1, 7, 5]

# data[::-1]

#0(n)

n = len(data)
reversed_data = []
for i in range(n-1, -1, -1):
    reversed_data.append(data[i])
print(f'The raversed data is :', reversed_data)
    

```

```python
data = [4, 6, 2, 3, 1, 7, 5]

n = len(data) - 1 
reversed_data2 = []
while n >= 0:
    reversed_data2.append(data[n])
    n -= 1
print(f'The raversed data is :', reversed_data2)
```
